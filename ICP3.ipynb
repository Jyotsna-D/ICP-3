Find the number of the columns in the data frame

```python
num_columns = df.shape[1]
print(num_columns)
```

Find the mean of the house prices

```python
mean_price = df['price'].mean()
print(mean_price)
```

Find the median of the house prices

```python
median_price = df['price'].median()
print(median_price)
```

Find the different types of property in the dataset

```python
property_types = df['type'].unique()
print(property_types)
```

Find the maximum price of a "condo" in the dataset

```python
max_condo_price = df[df['type'].str.lower() == 'condo']['price'].max()
print(max_condo_price)
```

Find the mean price of each type of property in the dataset

```python
mean_price_by_type = df.groupby('type')['price'].mean()
print(mean_price_by_type)
```

Find the count of different property types in dataset

```python
count_by_type = df['type'].value_counts()
print(count_by_type)
```
